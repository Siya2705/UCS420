<h2 style="color:#245B45;"><b>QUESTION 1: TEXT PREPROCESSING AND TOKENIZATION</b></h2>

**Write a unique paragraph (5-6 sentences) about your favorite topic (e.g., sports, technology, food, books, etc.).**

**1.** Convert text to lowercase and remove punctuation using `re`.

**2.** Tokenize the text into words and sentences.

**3.** Split using `split()` and `word_tokenize()` and compare how Python split and NLTK’s `word_tokenize()` differ.

**4.** Remove stopwords (using NLTK's stopwords list).

**5.** Display word frequency distribution (excluding stopwords).

```python
import re
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.probability import FreqDist

text = """Birds are fascinating creatures that make nature beautiful with their colours and songs.
They play an important role in maintaining the balance of our ecosystem.
Some birds help in pollination, while others control insect populations and spread seeds.
I enjoy watching birds flying freely in the sky and building nests on trees.
Different species of birds have unique features, habitats, and behaviours.
Protecting birds and their natural habitats is essential for maintaining biodiversity."""

clean_text = re.sub(r'[^\w\s]', '', text.lower())

print("Cleaned Text:\n", clean_text)

words = word_tokenize(clean_text)
sentences = sent_tokenize(text)

print("\nWord Tokens:\n", words)
print("\nSentence Tokens:\n", sentences)

split_words = clean_text.split()
nltk_words = word_tokenize(clean_text)

print("\nUsing split():\n", split_words)
print("\nUsing word_tokenize():\n", nltk_words)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in nltk_words if word not in stop_words]

print("\nAfter Stopword Removal:\n", filtered_words)

freq = FreqDist(filtered_words)

print("\nWord Frequencies:\n", freq)

freq.plot(10, title="Birds Word Frequency Distribution",color="g")
plt.show()
```

<h2 style="color:#245B45;"><b>QUESTION 2: STEMMING AND LEMMATIZATION</b></h2>

**Using the same paragraph from Q1:**

**1.** Extract all words with only alphabets using `re.findall()`.

**2.** Remove stop words using NLTK’s stopword list.

**3.** Perform stemming with `PorterStemmer`.

**4.** Perform lemmatization with `WordNetLemmatizer`.

**5.** Compare the stemmed and lemmatized outputs and explain when you’d prefer one over the other.

```python
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

text = """Birds are fascinating creatures that make nature beautiful with their colours and songs.
They play an important role in maintaining the balance of our ecosystem.
Some birds help in pollination, while others control insect populations and spread seeds.
I enjoy watching birds flying freely in the sky and building nests on trees.
Different species of birds have unique features, habitats, and behaviours.
Protecting birds and their natural habitats is essential for maintaining biodiversity."""

words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

print("Extracted Words:\n", words)

stop_words = set(stopwords.words('english'))
filtered_words = [word for word in words if word not in stop_words]

print("\nAfter Stopword Removal:\n", filtered_words)

ps = PorterStemmer()
stemmed_words = [ps.stem(word) for word in filtered_words]

print("\nStemmed Words:\n", stemmed_words)

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("\nLemmatized Words:\n", lemmatized_words)
```

<h2 style="color:#2E8B57;"><b>QUESTION 3: BAG OF WORDS AND TF-IDF</b></h2>

**Choose 3 short texts of your own (e.g., different news headlines, product reviews).**

**1.** Use `CountVectorizer` to generate the Bag of Words representation.

**2.** Use `TfidfVectorizer` to compute TF-IDF scores.

**3.** Print and interpret the top 3 keywords from each text using TF-IDF.

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

texts = [
    "Peacocks are beautiful birds with colourful feathers and long tails.",
    "Parrots are intelligent birds that can imitate human speech and sounds.",
    "Eagles are powerful birds known for their sharp eyesight and hunting skills."
]

count_vectorizer = CountVectorizer()
bow_matrix = count_vectorizer.fit_transform(texts)

print("Bag of Words Vocabulary:")
print(count_vectorizer.get_feature_names_out())

print("\nBag of Words Representation:")
print(bow_matrix.toarray())

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(texts)

feature_names = tfidf_vectorizer.get_feature_names_out()

print("\nTF-IDF Scores:")
print(tfidf_matrix.toarray())

for i, text in enumerate(texts):
    scores = tfidf_matrix[i].toarray()[0]
    top_indices = scores.argsort()[-3:][::-1]

    print("\nText", i+1, ":", text)
    print("Top 3 Keywords:")

    for index in top_indices:
        print(feature_names[index], ":", round(scores[index], 3))
```

<h2 style="color:#2E8B57;"><b>QUESTION 4: JACCARD SIMILARITY AND COSINE SIMILARITY</b></h2>

**Write 2 short texts (4–6 lines each) describing two different technologies (e.g., AI vs Blockchain).**

**1.** Preprocess and tokenize both texts.

**2.** Calculate:

&nbsp;&nbsp;&nbsp;&nbsp;**a.** Jaccard Similarity using sets.

&nbsp;&nbsp;&nbsp;&nbsp;**b.** Cosine Similarity using `TfidfVectorizer` + `cosine_similarity()`.

&nbsp;&nbsp;&nbsp;&nbsp;**c.** Analyze which similarity metric gives better insights in your case.

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

text1 = """Artificial Intelligence enables machines to perform tasks that require human intelligence.
AI systems can learn from data and make predictions.
They are used in healthcare, education, robotics and automation.
Machine learning is an important part of artificial intelligence.
AI improves efficiency and supports decision making."""

text2 = """Blockchain is a distributed technology used to store information securely.
It maintains records across multiple computers.
Blockchain provides transparency and prevents data manipulation.
It is widely used in cryptocurrency and digital transactions.
Smart contracts are an important feature of blockchain."""

clean_text1 = re.sub(r'[^\w\s]', '', text1.lower())
clean_text2 = re.sub(r'[^\w\s]', '', text2.lower())

words1 = clean_text1.split()
words2 = clean_text2.split()

print("Text 1 Tokens:\n", words1)
print("\nText 2 Tokens:\n", words2)

set1 = set(words1)
set2 = set(words2)

jaccard = len(set1.intersection(set2)) / len(set1.union(set2))

print("\nJaccard Similarity:", jaccard)

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform([clean_text1, clean_text2])

cosine = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])

print("\nCosine Similarity:", cosine[0][0])
```

<h2 style="color:#2E8B57;"><b>QUESTION 5: SENTIMENT ANALYSIS AND WORD CLOUD</b></h2>

**Write a short review for a product or service.**

**1.** Use TextBlob or VADER to find polarity and subjectivity for each review.

**2.** Classify reviews into Positive / Negative / Neutral.

**3.** Create a word cloud using the `wordcloud` library for all positive reviews.

```python
from textblob import TextBlob
from wordcloud import WordCloud
import matplotlib.pyplot as plt

reviews = [
    "This bird seed mix is excellent. My parrots love it and the seeds are fresh.",
    "Amazing quality bird food. The birds enjoy eating it every day.",
    "The seed mix is terrible. It smells bad and the quality is very poor.",
    "I am disappointed with this product. The seeds were stale and insects were present.",
    "The bird seed mix is average. It is neither very good nor very bad."
]

positive_reviews = []

for review in reviews:
    analysis = TextBlob(review)

    polarity = analysis.sentiment.polarity
    subjectivity = analysis.sentiment.subjectivity

    if polarity > 0:
        sentiment = "Positive"
        positive_reviews.append(review)
    elif polarity < 0:
        sentiment = "Negative"
    else:
        sentiment = "Neutral"

    print("\nReview:", review)
    print("Polarity:", round(polarity, 3))
    print("Subjectivity:", round(subjectivity, 3))
    print("Sentiment:", sentiment)

positive_text = " ".join(positive_reviews)

wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    colormap='summer'
).generate(positive_text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title("Word Cloud of Positive Bird Seed Reviews")
plt.show()
```

<h2 style="color:#2E8B57;"><b>QUESTION 6: TEXT GENERATION USING LSTM</b></h2>

**Choose your own paragraph (~100 words) as training data.**

**1.** Tokenize text using `Tokenizer()` from `keras.preprocessing.text`.

**2.** Create input sequences and build a simple LSTM or Dense model.

**3.** Train the model and generate 2–3 new lines of text starting from any seed word you provide.

```python
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

text = """Birds are beautiful creatures found in nature.
They have different colours, shapes and sizes.
Parrots are intelligent birds that can imitate human sounds.
Peacocks are famous for their colourful feathers.
Eagles have sharp eyesight and are excellent hunters.
Sparrows build small nests on trees and buildings.
Birds play an important role in maintaining the ecosystem.
Some birds help in spreading seeds and controlling insects.
Many people enjoy watching birds in gardens and forests.
We should protect birds and their natural habitats."""

tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

input_sequences = []

for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i+1]
        input_sequences.append(n_gram_sequence)

max_sequence_len = max([len(x) for x in input_sequences])

input_sequences = np.array(
    pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')
)

X = input_sequences[:, :-1]
y = input_sequences[:, -1]

model = Sequential()
model.add(Embedding(total_words, 10))
model.add(LSTM(50))
model.add(Dense(total_words, activation='softmax'))

model.compile(loss='sparse_categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

model.fit(X, y, epochs=100, verbose=1)

seed_text = "birds"

for _ in range(3):
    token_list = tokenizer.texts_to_sequences([seed_text])[0]
    token_list = pad_sequences([token_list], maxlen=max_sequence_len-1, padding='pre')

    predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]

    output_word = ""

    for word, index in tokenizer.word_index.items():
        if index == predicted:
            output_word = word
            break

    seed_text += " " + output_word

print("\nGenerated Text:\n", seed_text)
```
